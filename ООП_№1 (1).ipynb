### Описание
В данной работе реализованы два базовых класса для системы логистики:
1. `Shipment` (Груз) — с атрибутом **вес**.
2. `Recipient` (Получатель) — с атрибутом **адрес**.

### Особенности реализации
- Использован конструктор `__init__` для инициализации атрибутов.
- Реализован метод `display_info()` для подробного вывода.
- Переопределен магический метод `__str__` для красивого вывода (Vibe Code).
- Код соответствует стандарту PEP 8 и содержит документацию (docstrings).

```python
# ==============================================================================
# Лабораторная работа 4.1. Вариант 7: Логистика
# ==============================================================================

class Shipment:
    """Класс, описывающий груз (отправление) в логистике."""

    def __init__(self, shipment_id: int, weight: float):
        """Конструктор: сохраняет ID груза и его вес."""
        self.shipment_id = shipment_id
        self.weight = weight  # Атрибут экземпляра из задания (вес)

    def display_info(self):    # Метод экземпляра
        """Выводит подробную информацию о грузе."""
        print(f"--- Груз №{self.shipment_id} ---")
        print(f"Вес отправления: {self.weight} кг")
        print("-" * 25)

    def __str__(self):
        """Vibe Code: красивая строка для функции print()"""
        return f" Груз #{self.shipment_id} | Вес: {self.weight} кг"


class Recipient:
    """Класс, описывающий получателя груза."""

    def __init__(self, name: str, address: str):
        """Конструктор: сохраняет имя получателя и его адрес."""
        self.name = name
        self.address = address  # Атрибут экземпляра из задания (адрес)

    def display_info(self):
        """Выводит подробную информацию о получателе."""
        print(f"--- Получатель: {self.name} ---")
        print(f"Адрес доставки: {self.address}")
        print("-" * 30)

    def __str__(self):
        """Vibe Code: красивая строка для функции print()"""
        return f" Получатель: {self.name} | Адрес: {self.address}"


# ==============================================================================
# ДЕМОНСТРАЦИЯ РАБОТЫ
# ==============================================================================
if __name__ == "__main__":
    print("=== ЗАПУСК ЛАБОРАТОРНОЙ РАБОТЫ (Вариант 7) ===\n")

    # 1. Создаем 3 объекта класса Shipment (Груз)
    cargo_1 = Shipment(shipment_id=101, weight=25.5)
    cargo_2 = Shipment(shipment_id=102, weight=150.0)
    cargo_3 = Shipment(shipment_id=103, weight=4.2)

    # 2. Создаем 3 объекта класса Recipient (Получатель)
    person_1 = Recipient(name="Иван Иванов", address="г. Москва, ул. Ленина, д. 1")
    person_2 = Recipient(name="ООО 'Ромашка'", address="г. Санкт-Петербург, Невский пр., д. 10")
    person_3 = Recipient(name="Анна Петрова", address="г. Казань, ул. Баумана, д. 5")

    # 3. Выводим информацию через print() (это проверит метод __str__)
    print(" Вывод через print() (метод __str__):")
    print(cargo_1)
    print(cargo_2)
    print(cargo_3)
    print() # пустая строка для красоты
    print(person_1)
    print(person_2)
    print(person_3)
    print("\n")

    # 4. Вывод через метод display_info()
    print(" Вывод через метод display_info():")
    print()

# Все грузы
cargo_1.display_info()
cargo_2.display_info()
cargo_3.display_info()

# Все получатели
person_1.display_info()
person_2.display_info()
person_3.display_info()
```
