```python
def get_element(lst, index):
    """Вернуть элемент списка по индексу.

    Параметры:
        - lst (list): список элементов
        - index (int): индекс элемента

    Результат:
        - элемент списка по указанному индексу

    Исключения:
        - IndexError: если индекс выходит за границы списка
        - TypeError: если переданные аргументы имеют неверный тип
    """
    try:
        return lst[index]
    except IndexError:
        raise IndexError(f"Индекс {index} выходит за границы списка (0..{len(lst)-1})")
    except TypeError:
        raise TypeError("Индекс должен быть целым числом")

def main():
    my_list = [1, 2, 3, 4, 5]

    try:
        idx_input = input("Введите индекс: ")
        if not idx_input.lstrip('-').isdigit():
            raise TypeError("Индекс должен быть целым числом")
        idx = int(idx_input)
        result = get_element(my_list, idx)
        print(f"Элемент: {result}")
    except IndexError as err:
        print(f"Ошибка индекса: {err}")
    except TypeError as err:
        print(f"Ошибка типа: {err}")
    except ValueError as err:
        print(f"Ошибка преобразования: {err}. Введите целое число.")
    except Exception as err:
        print(f"Непредвиденная ошибка: {err}")

if __name__ == "__main__":
    main()


    #Пояснение исправлений:

#Строка 3: Исходная функция get_element не обрабатывала возможные ошибки при обращении к списку (выход индекса за границы, неверный тип индекса).

#Строка 8-9: В функции main не было обработки ошибок при преобразовании введённой строки в целое число (возможен ValueError).

#Строка 11: Отсутствовала обработка исключений при вызове get_element.

#Добавлена проверка ввода: Перед преобразованием в int проверяется, что введённая строка является корректным целым числом (с учётом возможного знака минус).
```
